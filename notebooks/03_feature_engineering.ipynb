```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/cleaned/campaign_data_cleaned.csv")
```

# Click Through Rate

```python
df['ctr'] = (
    df['clicks'] /
    df['impressions']
) * 100
```

# Conversion Rate

```python
df['conversion_rate'] = np.where(
    df['clicks'] == 0,
    0,
    (df['conversions'] / df['clicks']) * 100
)
```

# Lead Conversion Rate

```python
df['lead_conversion_rate'] = np.where(
    df['leads'] == 0,
    0,
    (df['conversions'] / df['leads']) * 100
)
```

# Engagement Rate

```python
df['engagement_rate'] = (
    (
        df['likes'] +
        df['shares'] +
        df['comments']
    )
    /
    df['impressions']
) * 100
```

# Cost Per Click

```python
df['cpc'] = np.where(
    df['clicks'] == 0,
    0,
    df['spend'] / df['clicks']
)
```

# Cost Per Lead

```python
df['cpl'] = np.where(
    df['leads'] == 0,
    0,
    df['spend'] / df['leads']
)
```

# Cost Per Conversion

```python
df['cpa'] = np.where(
    df['conversions'] == 0,
    0,
    df['spend'] / df['conversions']
)
```

# Revenue Per Click

```python
df['revenue_per_click'] = np.where(
    df['clicks'] == 0,
    0,
    df['revenue'] / df['clicks']
)
```

# Revenue Per Conversion

```python
df['revenue_per_conversion'] = np.where(
    df['conversions'] == 0,
    0,
    df['revenue'] / df['conversions']
)
```

# Profit Margin

```python
df['profit_margin'] = np.where(
    df['revenue'] == 0,
    0,
    (df['profit'] / df['revenue']) * 100
)
```

# Engagement Per Session

```python
df['engagement_per_session'] = (
    (
        df['likes'] +
        df['shares'] +
        df['comments']
    )
    /
    df['session_duration_sec']
)
```

# Customer Satisfaction Score

```python
df['customer_rating_pct'] = (
    df['customer_rating'] / 5
) * 100
```

# Date Based Features

```python
df['campaign_date'] = pd.to_datetime(
    df['campaign_date']
)
```

```python
# Month
df['month'] = df['campaign_date'].dt.month
```

```python
# Quarter
df['quarter'] = df['campaign_date'].dt.quarter
```

```python
# Days of week
df['day_of_week'] = (
    df['campaign_date']
    .dt.day_name()
)
```

```python
# Weekend
df['is_weekend'] = (
    df['campaign_date']
    .dt.dayofweek >= 5
).astype(int)
```

```python
df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)
```
